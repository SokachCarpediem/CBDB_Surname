```python
def build_bjx_raw(db, seq, s2t, out_path=None):
    # 《百家姓》序位表：只保留简体+繁体两种字形，异体字暂时先不合并
    rows = []
    for i, c in enumerate(seq, 1):
        sc = c                        # 简体写法
        tc = s2t.get(c, c)            # 繁体写法（无映射则同形）
        if sc == tc:
            n = db.get(sc, 0)         # 简繁同形，只加一次
        else:
            n = db.get(sc, 0) + db.get(tc, 0)   # 简+繁都计入该序位
        rows.append([tc, n, i])
    if out_path:
        with open(out_path, "w", encoding="utf-8") as f:
            json.dump(rows, f, ensure_ascii=False, indent=1)
    return rows

bjx = build_bjx_raw(db, seq, S2T, BJX_OUT)
print("BJX 行数:", len(bjx))
print("有數據  :", sum(1 for x in bjx if x[1] > 0), "| 無數據記0:", sum(1 for x in bjx if x[1] == 0))
print("前 8 位:", bjx[:8])
print("saved:", BJX_OUT)
```

```python
def build_extra_raw(db, seq, s2t, out_path=None):
    #非《百家姓》表：排除百家姓的简体+繁体两形，其余按人数降序。
    bjx_forms = set()
    for c in seq:
        bjx_forms.add(c)
        bjx_forms.add(s2t.get(c, c))
    extra = [[s, n] for s, n in db.items() if s not in bjx_forms]
    extra.sort(key=lambda x: (-x[1], x[0]))
    if out_path:
        with open(out_path, "w", encoding="utf-8") as f:
            json.dump(extra, f, ensure_ascii=False, indent=1)
    return extra

extra = build_extra_raw(db, seq, S2T, EXTRA_OUT)
print("EXTRA 行数:", len(extra))
print("人数      :", sum(n for _, n in extra))
print("top 10:", extra[:10])
print("saved:", EXTRA_OUT)
```
